```python
# ============================================================
# FINAL NOTEBOOK-INLINE EVALUATION ANALYSIS
# Corrected logical round parsing:
#   - 3 agent messages = 1 logical round
#   - keep only round_0 ... round_4
#   - round_0 can be SC from io_output
# ============================================================

import os
import re
import json
import glob
from pathlib import Path
from collections import Counter

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

try:
    from IPython.display import display
except Exception:
    display = print


# ============================================================
# 0. Configuration
# ============================================================

BASE_DIR = "."

DIR_CONFIG = {
    "MAD": os.path.join(BASE_DIR, "MAD", "*.json"),
    "DMAD": os.path.join(BASE_DIR, "DMAD", "*.json"),
    "Reasoning": os.path.join(BASE_DIR, "reasoning", "*.json"),
    "Vote": os.path.join(BASE_DIR, "outputs_vote", "*.json"),
}

N_AGENTS = 3
MAX_LOGICAL_ROUNDS = 5          # keep round_0, round_1, round_2, round_3, round_4
USE_IO_OUTPUT_AS_ROUND0 = True  # recommended: round_0 = SC / initial independent answers

MAX_EXAMPLES_PER_ERROR_TYPE = 8

LABEL_REGEX_LIST = [
    re.compile(r"Label\s*:\s*([01])", re.IGNORECASE),
    re.compile(r'"label"\s*:\s*([01])', re.IGNORECASE),
    re.compile(r"'label'\s*:\s*([01])", re.IGNORECASE),
]


# ============================================================
# 1. General Helpers
# ============================================================

def safe_load_json(path):
    with open(path, "r", encoding="utf-8") as f:
        return json.load(f)


def flatten_texts(obj):
    texts = []

    if obj is None:
        return texts

    if isinstance(obj, str):
        texts.append(obj)

    elif isinstance(obj, list):
        for x in obj:
            texts.extend(flatten_texts(x))

    elif isinstance(obj, dict):
        for v in obj.values():
            texts.extend(flatten_texts(v))

    else:
        texts.append(str(obj))

    return texts


def extract_all_labels(text):
    if not isinstance(text, str):
        return []

    labels = []

    for rgx in LABEL_REGEX_LIST:
        matches = rgx.findall(text)
        for m in matches:
            try:
                labels.append(int(m))
            except Exception:
                pass

    return labels


def extract_last_label(text):
    labels = extract_all_labels(text)
    if len(labels) == 0:
        return None
    return labels[-1]


def majority_vote(labels, threshold=None):
    labels = [x for x in labels if x in [0, 1]]

    if len(labels) == 0:
        return None

    pos = sum(labels)

    if threshold is None:
        threshold = int(np.ceil(len(labels) / 2))

    return 1 if pos >= threshold else 0


def compute_binary_metrics(y_true, y_pred):
    y_true = pd.Series(y_true)
    y_pred = pd.Series(y_pred)

    mask = y_true.isin([0, 1]) & y_pred.isin([0, 1])
    y_true = y_true[mask].astype(int)
    y_pred = y_pred[mask].astype(int)

    if len(y_true) == 0:
        return None

    tp = ((y_true == 1) & (y_pred == 1)).sum()
    fp = ((y_true == 0) & (y_pred == 1)).sum()
    fn = ((y_true == 1) & (y_pred == 0)).sum()
    tn = ((y_true == 0) & (y_pred == 0)).sum()

    precision = tp / (tp + fp + 1e-9)
    recall = tp / (tp + fn + 1e-9)
    f1 = 2 * precision * recall / (precision + recall + 1e-9)
    acc = (tp + tn) / (tp + fp + fn + tn + 1e-9)

    return {
        "P": precision,
        "R": recall,
        "F1": f1,
        "Acc": acc,
        "TP": tp,
        "FP": fp,
        "FN": fn,
        "TN": tn,
        "N": len(y_true),
    }


def infer_method_name(path):
    return Path(path).stem


def infer_prompt_family(file_name):
    name = file_name.lower()

    tags = []

    candidates = [
        "allio",
        "io-cot-sbp",
        "posneg",
        "carp_json",
        "dtg_json",
        "io_pos",
        "ccot",
        "ddcot",
        "carp",
        "dtg",
        "cot",
        "sbp",
        "kg",
        "io",
    ]

    for key in candidates:
        if key in name:
            tags.append(key)

    if len(tags) == 0:
        return "unknown"

    return "+".join(tags)


def get_output_keys(sample):
    keys = []

    for k, v in sample.items():
        if k.endswith("_output") or k == "outputs":
            if isinstance(v, (list, str, dict)):
                keys.append(k)

    return keys


def extract_outputs_from_sample(sample):
    output_texts = []
    output_keys = get_output_keys(sample)

    for k in output_keys:
        v = sample.get(k)

        if isinstance(v, list):
            for item in v:
                if isinstance(item, str):
                    output_texts.append(item)
                else:
                    output_texts.extend(flatten_texts(item))

        elif isinstance(v, str):
            output_texts.append(v)

        elif isinstance(v, dict):
            output_texts.extend(flatten_texts(v))

    output_labels = [extract_last_label(x) for x in output_texts]
    output_labels = [x for x in output_labels if x in [0, 1]]

    return output_texts, output_labels, output_keys


def extract_agent_messages_flat(sample):
    """
    Flatten agent_contexts into a single ordered list of messages.
    This is the key fix.

    Instead of treating every message as one round, we later group:
        every 3 messages = one logical round
    """
    agent_contexts = sample.get("agent_contexts", None)

    if not isinstance(agent_contexts, list) or len(agent_contexts) == 0:
        return []

    messages = []

    for block in agent_contexts:
        if isinstance(block, list):
            for msg in block:
                texts = flatten_texts(msg)
                if len(texts) > 0:
                    messages.append("\n".join(texts))
        else:
            texts = flatten_texts(block)
            if len(texts) > 0:
                messages.append("\n".join(texts))

    return messages


def extract_logical_rounds(sample, output_labels):
    """
    Correct logical round extraction.

    Two modes:

    1. USE_IO_OUTPUT_AS_ROUND0 = True:
       round_0 = majority vote over io_output labels
       agent_contexts grouped by 3 messages become round_1 ... round_4

    2. USE_IO_OUTPUT_AS_ROUND0 = False:
       agent_contexts grouped by 3 messages become round_0 ... round_4

    In all cases, only round_0 ... round_4 are kept.
    """
    round_preds = {}

    next_round_idx = 0

    # ---- round_0 as SC baseline ----
    if USE_IO_OUTPUT_AS_ROUND0 and len(output_labels) > 0:
        round_preds["round_0"] = majority_vote(output_labels)
        next_round_idx = 1

    # ---- debate messages grouped by N_AGENTS ----
    messages = extract_agent_messages_flat(sample)

    chunks = [
        messages[i:i + N_AGENTS]
        for i in range(0, len(messages), N_AGENTS)
    ]

    for chunk in chunks:
        if next_round_idx >= MAX_LOGICAL_ROUNDS:
            break

        labels_this_round = []

        for msg in chunk:
            lab = extract_last_label(msg)
            if lab in [0, 1]:
                labels_this_round.append(lab)

        if len(labels_this_round) > 0:
            round_preds[f"round_{next_round_idx}"] = majority_vote(labels_this_round)
            next_round_idx += 1

    return round_preds


def extract_usage(sample):
    usage = sample.get("usage", {})

    if not isinstance(usage, dict):
        usage = {}

    return {
        "prompt_tokens": usage.get("prompt_tokens", np.nan),
        "completion_tokens": usage.get("completion_tokens", np.nan),
        "total_tokens": usage.get("total_tokens", np.nan),
    }


def avg_output_length(texts):
    if not isinstance(texts, list) or len(texts) == 0:
        return np.nan
    return np.mean([len(str(t).split()) for t in texts])


# ============================================================
# 2. Parse One JSON File
# ============================================================

def parse_file(path, group_name):
    method = infer_method_name(path)
    prompt_family = infer_prompt_family(method)

    raw = safe_load_json(path)

    rows = []
    field_counter = Counter()
    output_key_counter = Counter()
    n_output_labels_list = []
    n_rounds_list = []
    n_agent_messages_list = []

    for i, sample in enumerate(raw):

        if not isinstance(sample, dict):
            continue

        for k in sample.keys():
            field_counter[k] += 1

        gold = sample.get("label", None)
        raw_id = sample.get("raw_id", sample.get("_raw_id", sample.get("num", i)))
        review_text = sample.get("text", "")

        usage = extract_usage(sample)

        output_texts, output_labels, output_keys = extract_outputs_from_sample(sample)

        for k in output_keys:
            output_key_counter[k] += 1

        agent_messages = extract_agent_messages_flat(sample)
        round_preds = extract_logical_rounds(sample, output_labels)

        n_output_labels_list.append(len(output_labels))
        n_rounds_list.append(len(round_preds))
        n_agent_messages_list.append(len(agent_messages))

        sc_pred = round_preds.get("round_0", None)

        if len(round_preds) > 0:
            last_round_key = sorted(
                round_preds.keys(),
                key=lambda x: int(x.split("_")[1])
            )[-1]
            final_pred = round_preds[last_round_key]
        elif len(output_labels) > 0:
            final_pred = majority_vote(output_labels)
        else:
            final_pred = None

        row = {
            "group": group_name,
            "method": method,
            "prompt_family": prompt_family,
            "raw_id": raw_id,
            "gold": gold,
            "text": review_text,
            "n_output_labels": len(output_labels),
            "n_agent_messages": len(agent_messages),
            "n_rounds": len(round_preds),
            "sc_pred": sc_pred,
            "final_pred": final_pred,
            "output_labels": output_labels,
            "output_texts": output_texts,
            "avg_output_words": avg_output_length(output_texts),
            "prompt_tokens": usage["prompt_tokens"],
            "completion_tokens": usage["completion_tokens"],
            "total_tokens": usage["total_tokens"],
        }

        for rk, rv in round_preds.items():
            row[rk] = rv

        rows.append(row)

    df = pd.DataFrame(rows)

    field_report = {
        "group": group_name,
        "file": method,
        "n_samples": len(df),
        "fields": ", ".join(sorted(field_counter.keys())),
        "output_keys": ", ".join(sorted(output_key_counter.keys())),
        "avg_n_output_labels": float(np.nanmean(n_output_labels_list)) if len(n_output_labels_list) else 0,
        "max_n_output_labels": max(n_output_labels_list) if len(n_output_labels_list) else 0,
        "avg_agent_messages": float(np.nanmean(n_agent_messages_list)) if len(n_agent_messages_list) else 0,
        "max_agent_messages": max(n_agent_messages_list) if len(n_agent_messages_list) else 0,
        "avg_n_rounds": float(np.nanmean(n_rounds_list)) if len(n_rounds_list) else 0,
        "max_n_rounds": max(n_rounds_list) if len(n_rounds_list) else 0,
        "prompt_family": prompt_family,
    }

    return df, field_report


# ============================================================
# 3. Parse All Files
# ============================================================

all_sample_dfs = []
field_reports = []

for group_name, pattern in DIR_CONFIG.items():
    files = sorted(glob.glob(pattern))
    print(f"\n[{group_name}] Found {len(files)} files")

    for path in files:
        try:
            df_file, report = parse_file(path, group_name)
            all_sample_dfs.append(df_file)
            field_reports.append(report)

            print(
                f"  OK: {Path(path).name} | "
                f"samples={report['n_samples']} | "
                f"max_outputs={report['max_n_output_labels']} | "
                f"max_agent_messages={report['max_agent_messages']} | "
                f"max_rounds={report['max_n_rounds']} | "
                f"outputs={report['output_keys']}"
            )

        except Exception as e:
            print(f"  FAILED: {Path(path).name} | {e}")

if len(all_sample_dfs) == 0:
    raise RuntimeError("No JSON files parsed. Check your folder paths.")

samples_df = pd.concat(all_sample_dfs, ignore_index=True)
field_report_df = pd.DataFrame(field_reports)

print("\n===== FIELD / ROUND / PROMPT REPORT =====")
display(field_report_df)


# ============================================================
# 4. Overall Metrics
# ============================================================

def metrics_for_group(df, pred_col):
    if pred_col not in df.columns:
        return None

    df_eval = df.dropna(subset=["gold", pred_col]).copy()

    df_eval = df_eval[
        df_eval["gold"].isin([0, 1]) &
        df_eval[pred_col].isin([0, 1])
    ]

    if len(df_eval) == 0:
        return None

    return compute_binary_metrics(df_eval["gold"], df_eval[pred_col])


overall_rows = []

for (group, method), g in samples_df.groupby(["group", "method"]):
    m = metrics_for_group(g, "final_pred")

    if m is None:
        continue

    overall_rows.append({
        "group": group,
        "method": method,
        "prompt_family": g["prompt_family"].iloc[0],
        "setting": "final_pred",
        "P": m["P"],
        "R": m["R"],
        "F1": m["F1"],
        "Acc": m["Acc"],
        "TP": m["TP"],
        "FP": m["FP"],
        "FN": m["FN"],
        "TN": m["TN"],
        "N": m["N"],
        "avg_total_tokens": g["total_tokens"].mean(),
        "total_tokens": g["total_tokens"].sum(),
    })

overall_df = pd.DataFrame(overall_rows).sort_values("F1", ascending=False)

print("\n===== OVERALL METRICS =====")
display(overall_df)


# ============================================================
# 5. Round Metrics
# ============================================================

round_cols_all = sorted(
    [c for c in samples_df.columns if re.match(r"round_\d+", c)],
    key=lambda x: int(x.split("_")[1])
)

# Hard safety: keep only round_0 ... round_4
round_cols_all = [
    c for c in round_cols_all
    if int(c.split("_")[1]) < MAX_LOGICAL_ROUNDS
]

round_rows = []

for (group, method), g in samples_df.groupby(["group", "method"]):

    for rc in round_cols_all:

        if rc not in g.columns:
            continue

        m = metrics_for_group(g, rc)

        if m is None:
            continue

        round_rows.append({
            "group": group,
            "method": method,
            "prompt_family": g["prompt_family"].iloc[0],
            "round": rc,
            "round_idx": int(rc.split("_")[1]),
            "P": m["P"],
            "R": m["R"],
            "F1": m["F1"],
            "Acc": m["Acc"],
            "N": m["N"],
        })

round_df = pd.DataFrame(round_rows)

print("\n===== ROUND METRICS =====")
if len(round_df) > 0:
    display(round_df.sort_values(["method", "round_idx"]))
else:
    print("No round metrics detected.")


# ============================================================
# 6. SC vs Best Debate
# ============================================================

sc_best_rows = []

if len(round_df) > 0:

    for (group, method), g in round_df.groupby(["group", "method"]):

        if not (g["round_idx"] == 0).any():
            continue

        sc = g[g["round_idx"] == 0].iloc[0]
        debate_rounds = g[g["round_idx"] >= 1]

        if len(debate_rounds) == 0:
            continue

        best = debate_rounds.sort_values("F1", ascending=False).iloc[0]

        sc_best_rows.append({
            "group": group,
            "method": method,
            "prompt_family": g["prompt_family"].iloc[0],
            "SC_round": "round_0",
            "SC_P": sc["P"],
            "SC_R": sc["R"],
            "SC_F1": sc["F1"],
            "Best_Debate_Round": best["round"],
            "Best_P": best["P"],
            "Best_R": best["R"],
            "Best_F1": best["F1"],
            "Delta_F1_vs_SC": best["F1"] - sc["F1"],
            "Conclusion": "Debate helps" if best["F1"] > sc["F1"] else "SC better",
        })

sc_best_df = pd.DataFrame(sc_best_rows)

print("\n===== SC VS BEST DEBATE =====")
if len(sc_best_df) > 0:
    display(sc_best_df.sort_values("Delta_F1_vs_SC", ascending=False))
else:
    print("No SC vs Best Debate table generated.")


# ============================================================
# 7. Voting Threshold Analysis
# ============================================================

vote_rows = []

for (group, method), g in samples_df.groupby(["group", "method"]):

    max_outputs = int(g["n_output_labels"].max()) if len(g) else 0

    if max_outputs < 2:
        continue

    for threshold in range(1, max_outputs + 1):

        y_true = []
        y_pred = []

        for _, row in g.iterrows():
            labels = row["output_labels"]

            if not isinstance(labels, list) or len(labels) == 0:
                continue

            pred = majority_vote(labels, threshold=threshold)

            if pred is None or row["gold"] not in [0, 1]:
                continue

            y_true.append(row["gold"])
            y_pred.append(pred)

        if len(y_true) == 0:
            continue

        m = compute_binary_metrics(y_true, y_pred)

        vote_rows.append({
            "group": group,
            "method": method,
            "prompt_family": g["prompt_family"].iloc[0],
            "threshold": threshold,
            "P": m["P"],
            "R": m["R"],
            "F1": m["F1"],
            "Acc": m["Acc"],
            "N": m["N"],
        })

vote_df = pd.DataFrame(vote_rows)

print("\n===== VOTING THRESHOLD METRICS =====")
if len(vote_df) > 0:
    display(vote_df.sort_values(["method", "threshold"]))

    vote_best_df = (
        vote_df.sort_values("F1", ascending=False)
        .groupby(["group", "method"], as_index=False)
        .first()
        .rename(columns={
            "threshold": "Best_Threshold",
            "P": "Best_P",
            "R": "Best_R",
            "F1": "Best_F1",
            "Acc": "Best_Acc",
        })
    )

    print("\n===== BEST VOTING THRESHOLD =====")
    display(vote_best_df.sort_values("Best_F1", ascending=False))

else:
    vote_best_df = pd.DataFrame()
    print("No voting threshold table generated.")


# ============================================================
# 8. Prediction Flip Rate
# ============================================================

flip_rows = []

if len(round_cols_all) >= 2:

    for (group, method), g in samples_df.groupby(["group", "method"]):

        available_rounds = [
            c for c in round_cols_all
            if c in g.columns and g[c].notna().sum() > 0
        ]

        available_rounds = [
            c for c in available_rounds
            if int(c.split("_")[1]) < MAX_LOGICAL_ROUNDS
        ]

        for i in range(len(available_rounds) - 1):

            c1 = available_rounds[i]
            c2 = available_rounds[i + 1]

            valid = g[[c1, c2]].dropna()

            if len(valid) == 0:
                continue

            flip = (valid[c1] != valid[c2]).mean()

            flip_rows.append({
                "group": group,
                "method": method,
                "prompt_family": g["prompt_family"].iloc[0],
                "transition": f"{c1}->{c2}",
                "from_round": int(c1.split("_")[1]),
                "to_round": int(c2.split("_")[1]),
                "flip_rate": flip,
                "N": len(valid),
            })

flip_df = pd.DataFrame(flip_rows)

print("\n===== PREDICTION FLIP RATE =====")
if len(flip_df) > 0:
    display(flip_df.sort_values(["method", "from_round"]))
else:
    print("No flip-rate data generated.")


# ============================================================
# 9. Reasoning Length and Error Analysis
# ============================================================

analysis_df = samples_df.dropna(subset=["gold", "final_pred"]).copy()

analysis_df = analysis_df[
    analysis_df["gold"].isin([0, 1]) &
    analysis_df["final_pred"].isin([0, 1])
].copy()

analysis_df["correct"] = analysis_df["gold"] == analysis_df["final_pred"]

analysis_df["error_type"] = "Correct"
analysis_df.loc[
    (analysis_df["gold"] == 0) & (analysis_df["final_pred"] == 1),
    "error_type"
] = "FP"

analysis_df.loc[
    (analysis_df["gold"] == 1) & (analysis_df["final_pred"] == 0),
    "error_type"
] = "FN"

reasoning_summary = (
    analysis_df
    .groupby(["group", "method", "prompt_family", "error_type"])
    .agg(
        count=("raw_id", "count"),
        avg_output_words=("avg_output_words", "mean"),
        avg_total_tokens=("total_tokens", "mean"),
    )
    .reset_index()
)

print("\n===== REASONING LENGTH / ERROR ANALYSIS =====")
display(reasoning_summary)


# ============================================================
# 10. Qualitative Error Examples
# ============================================================

case_examples = []

for error_type in ["FP", "FN"]:

    subset = (
        analysis_df[analysis_df["error_type"] == error_type]
        .head(MAX_EXAMPLES_PER_ERROR_TYPE)
    )

    for _, row in subset.iterrows():

        outputs_preview = row["output_texts"][:2] if isinstance(row["output_texts"], list) else []

        case_examples.append({
            "error_type": error_type,
            "group": row["group"],
            "method": row["method"],
            "raw_id": row["raw_id"],
            "gold": row["gold"],
            "pred": row["final_pred"],
            "review_text": row["text"],
            "outputs_preview": outputs_preview,
        })

case_examples_df = pd.DataFrame(case_examples)

print("\n===== QUALITATIVE ERROR EXAMPLES =====")
if len(case_examples_df) > 0:
    display(case_examples_df)
else:
    print("No FP/FN examples found.")


# ============================================================
# 11. Inline Visualizations
# ============================================================

# ---- 11.1 Overall F1 comparison ----
if len(overall_df) > 0:

    plot_df = overall_df.sort_values("F1", ascending=False).head(15)

    plt.figure(figsize=(11, 5))
    plt.bar(plot_df["method"], plot_df["F1"])
    plt.xticks(rotation=75, ha="right")
    plt.ylabel("F1")
    plt.title("Overall F1 Comparison")
    plt.tight_layout()
    plt.show()


# ---- 11.2 Debate round F1 curves ----
if len(round_df) > 0:

    plt.figure(figsize=(10, 5))

    for method, g in round_df.groupby("method"):
        g = g.sort_values("round_idx")
        plt.plot(g["round_idx"], g["F1"], marker="o", label=method)

    plt.xlabel("Logical Debate Round")
    plt.ylabel("F1")
    plt.title("F1 Across Logical Debate Rounds")
    plt.legend(fontsize=7)
    plt.tight_layout()
    plt.show()


# ---- 11.3 Voting threshold F1 curves ----
if len(vote_df) > 0:

    plt.figure(figsize=(10, 5))

    for method, g in vote_df.groupby("method"):
        g = g.sort_values("threshold")
        plt.plot(g["threshold"], g["F1"], marker="o", label=method)

    plt.xlabel("Voting Threshold")
    plt.ylabel("F1")
    plt.title("F1 Across Voting Thresholds")
    plt.legend(fontsize=7)
    plt.tight_layout()
    plt.show()


# ---- 11.4 Prediction flip-rate curves ----
if len(flip_df) > 0:

    plt.figure(figsize=(10, 5))

    for method, g in flip_df.groupby("method"):
        g = g.sort_values("from_round")
        plt.plot(g["to_round"], g["flip_rate"], marker="o", label=method)

    plt.xlabel("Logical Debate Round")
    plt.ylabel("Prediction Flip Rate")
    plt.title("Prediction Flip Rate Across Logical Debate Rounds")
    plt.legend(fontsize=7)
    plt.tight_layout()
    plt.show()


# ---- 11.5 Reasoning length by error type ----
if len(reasoning_summary) > 0:

    pivot_len = (
        reasoning_summary
        .groupby("error_type")["avg_output_words"]
        .mean()
        .reset_index()
    )

    plt.figure(figsize=(6, 4))
    plt.bar(pivot_len["error_type"], pivot_len["avg_output_words"])
    plt.xlabel("Error Type")
    plt.ylabel("Average Output Words")
    plt.title("Reasoning Length by Error Type")
    plt.tight_layout()
    plt.show()


# ============================================================
# 12. Final Check
# ============================================================

print("\nDONE.")
print(f"Logical rounds are capped at: round_0 to round_{MAX_LOGICAL_ROUNDS - 1}")
print(f"Every {N_AGENTS} agent messages are treated as one logical round.")
print("All tables and figures are displayed inline. No files were saved.")
```

```python
import json
import re
import glob
from collections import Counter
import pandas as pd
import matplotlib.pyplot as plt


# ============================================================
# 1. 从输出文本中解析 Label
# ============================================================

def parse_label(text):
    if not isinstance(text, str):
        return None
    m = re.search(r"Label\s*:\s*([01])", text)
    return int(m.group(1)) if m else None


# ============================================================
# 2. 从单个文件加载 votes（不假设 3 还是 5）
# ============================================================

def load_votes_from_file(json_path):
    vote_matrix = []
    gold_labels = []

    with open(json_path, "r", encoding="utf-8") as f:
        data = json.load(f)

    for item in data:
        if "label" not in item or "outputs" not in item:
            continue

        votes = []
        for out in item["outputs"]:
            lab = parse_label(out)
            if lab is not None:
                votes.append(lab)

        if len(votes) > 0:
            vote_matrix.append(votes)
            gold_labels.append(item["label"])

    return vote_matrix, gold_labels


# ============================================================
# 3. 投票规则
# ============================================================

def majority_vote(votes):
    return Counter(votes).most_common(1)[0][0]

def threshold_vote(votes, k):
    return 1 if sum(votes) >= k else 0


# ============================================================
# 4. 对单个文件做阈值评估（k 由文件名决定）
# ============================================================

def evaluate_thresholds(vote_matrix, gold_labels, max_k):
    rows = []

    for k in range(1, max_k + 1):
        TP = FP = FN = TN = 0

        for votes, y in zip(vote_matrix, gold_labels):
            pred = threshold_vote(votes, k)

            if pred == 1 and y == 1:
                TP += 1
            elif pred == 1 and y == 0:
                FP += 1
            elif pred == 0 and y == 1:
                FN += 1
            else:
                TN += 1

        P = TP / (TP + FP) if TP + FP > 0 else 0
        R = TP / (TP + FN) if TP + FN > 0 else 0
        F1 = 2 * P * R / (P + R) if P + R > 0 else 0

        rows.append({
            "threshold": k,
            "P": P,
            "R": R,
            "F1": F1,
            "TP": TP,
            "FP": FP,
            "FN": FN,
            "TN": TN,
            "N": len(gold_labels)
        })

    return pd.DataFrame(rows)


# ============================================================
# 5. 遍历所有 JSON，按文件名决定 3 or 5 answers
# ============================================================

json_files = glob.glob("./resoning/*.json")

for path in json_files:
    votes, gold = load_votes_from_file(path)

    if not votes:
        continue

    filename = path.split("/")[-1]

    # ✅ 文件名规则：写了 5answers / -5 → 5 个回答
    if "5answers" in filename or filename.endswith("-5.json"):
        max_k = 5
        answer_type = "5 answers"
    else:
        max_k = 3
        answer_type = "3 answers"

    print(f"\nFile: {filename}")
    print(f"Samples: {len(gold)} | Mode: {answer_type}")

    df = evaluate_thresholds(votes, gold, max_k=max_k)
    display(df)

    plt.figure(figsize=(6, 4))
    plt.plot(df["threshold"], df["F1"], marker="o")
    plt.xlabel("Voting Threshold")
    plt.ylabel("F1")
    plt.title(f"{filename} ({answer_type})")
    plt.grid(True, alpha=0.3)
    plt.tight_layout()
    plt.show()
```

```python
import json
import re
import glob
from collections import Counter
import pandas as pd
import matplotlib.pyplot as plt


# ============================================================
# 1. Robust label parser（关键：兜底）
# ============================================================

def parse_label_robust(text):
    """
    Robustly extract binary label from messy outputs.
    Supports:
    - 'Label: 0/1'
    - JSON string: {"label": 0/1, ...}
    """
    if not isinstance(text, str):
        return None

    # Case 1: Label: X
    m = re.search(r"Label\s*:\s*([01])", text)
    if m:
        return int(m.group(1))

    # Case 2: JSON-like string
    text = text.strip()
    if text.startswith("{") and text.endswith("}"):
        try:
            obj = json.loads(text)
            if "label" in obj and obj["label"] in [0, 1]:
                return int(obj["label"])
        except Exception:
            pass

    return None


# ============================================================
# 2. Load votes from ONE file
# ============================================================

def load_votes_from_file(json_path):
    vote_matrix = []
    gold_labels = []

    with open(json_path, "r", encoding="utf-8") as f:
        data = json.load(f)

    for item in data:
        if "label" not in item or "outputs" not in item:
            continue

        votes = []
        for out in item["outputs"]:
            lab = parse_label_robust(out)
            if lab is not None:
                votes.append(lab)

        # 至少要有 1 票才算
        if len(votes) > 0:
            vote_matrix.append(votes)
            gold_labels.append(item["label"])

    return vote_matrix, gold_labels


# ============================================================
# 3. Voting rules
# ============================================================

def majority_vote(votes):
    return Counter(votes).most_common(1)[0][0]

def threshold_vote(votes, k):
    return 1 if sum(votes) >= k else 0


# ============================================================
# 4. Threshold evaluation
# ============================================================

def evaluate_thresholds(vote_matrix, gold_labels, max_k):
    rows = []

    for k in range(1, max_k + 1):
        TP = FP = FN = TN = 0

        for votes, y in zip(vote_matrix, gold_labels):
            pred = threshold_vote(votes, k)

            if pred == 1 and y == 1:
                TP += 1
            elif pred == 1 and y == 0:
                FP += 1
            elif pred == 0 and y == 1:
                FN += 1
            else:
                TN += 1

        P = TP / (TP + FP) if TP + FP > 0 else 0
        R = TP / (TP + FN) if TP + FN > 0 else 0
        F1 = 2 * P * R / (P + R) if P + R > 0 else 0

        rows.append({
            "threshold": k,
            "P": P,
            "R": R,
            "F1": F1,
            "TP": TP,
            "FP": FP,
            "FN": FN,
            "TN": TN,
            "N": len(gold_labels)
        })

    return pd.DataFrame(rows)


# ============================================================
# 5. Plot helper
# ============================================================

def plot_threshold_curve(df, title):
    plt.figure(figsize=(6, 4))
    plt.plot(df["threshold"], df["F1"], marker="o")
    plt.xlabel("Voting Threshold")
    plt.ylabel("F1")
    plt.title(title)
    plt.grid(True, alpha=0.3)
    plt.tight_layout()
    plt.show()


# ============================================================
# 6. MAIN: run ALL files automatically
# ============================================================

json_files = glob.glob("./resoning/*.json")

print("Found files:", len(json_files))

all_results = []

for path in json_files:
    votes, gold = load_votes_from_file(path)
    if not votes:
        continue

    filename = path.split("/")[-1]

    # ✅ 文件名规则（你已经确认）
    if "5answers" in filename or filename.endswith("-5.json"):
        max_k = 5
        mode = "5 answers"
    else:
        max_k = 3
        mode = "3 answers"

    print(f"\nFile: {filename}")
    print(f"Samples: {len(gold)} | Mode: {mode}")

    df = evaluate_thresholds(votes, gold, max_k)
    df["file"] = filename
    df["mode"] = mode

    display(df)
    plot_threshold_curve(df, f"{filename} ({mode})")

    all_results.append(df)


# ============================================================
# 7. Combined table (optional, but useful)
# ============================================================

if all_results:
    final_df = pd.concat(all_results, ignore_index=True)
    display(final_df)
```

```python
import re
import pandas as pd
from collections import defaultdict

# ---------- utilities ----------

LABEL_RE = re.compile(r"Label:\s*([01])")

def parse_label(text):
    if not isinstance(text, str):
        return None
    m = LABEL_RE.search(text)
    return int(m.group(1)) if m else None


def compute_f1(tp, fp, fn):
    p = tp / (tp + fp) if tp + fp > 0 else 0.0
    r = tp / (tp + fn) if tp + fn > 0 else 0.0
    f1 = 2 * p * r / (p + r) if p + r > 0 else 0.0
    return p, r, f1


# ---------- core logic ----------

def debate_results_from_raw(samples, vote_threshold=2):
    """
    Parameters
    ----------
    samples : list[dict]
        Each dict must contain:
        - label: ground truth (0/1)
        - answers_: List[List[str]]  # outer list = rounds, inner list = 3 agent outputs
    vote_threshold : int
        >= vote_threshold positive votes → predict 1

    Returns
    -------
    round_table : pd.DataFrame
        Per-round Precision / Recall / F1

    summary_table : pd.DataFrame
        SC_F1, Best_F1, Best_Round, Delta_F1
    """

    # round -> counts
    stats = defaultdict(lambda: {"TP": 0, "FP": 0, "FN": 0})

    for sample in samples:
        gt = sample["label"]
        rounds = sample["answers_"]  # List[List[str]]

        for r, agent_outputs in enumerate(rounds):
            votes = [parse_label(x) for x in agent_outputs]
            votes = [v for v in votes if v is not None]

            if len(votes) == 0:
                continue

            pred = 1 if sum(votes) >= vote_threshold else 0

            if pred == 1 and gt == 1:
                stats[r]["TP"] += 1
            elif pred == 1 and gt == 0:
                stats[r]["FP"] += 1
            elif pred == 0 and gt == 1:
                stats[r]["FN"] += 1
            # TN not needed for F1

    # ---------- build per-round table ----------
    round_rows = []
    f1_by_round = {}

    for r in sorted(stats):
        tp, fp, fn = stats[r]["TP"], stats[r]["FP"], stats[r]["FN"]
        p, r_, f1 = compute_f1(tp, fp, fn)

        round_rows.append({
            "Round": r,
            "TP": tp,
            "FP": fp,
            "FN": fn,
            "Precision": p,
            "Recall": r_,
            "F1": f1
        })

        f1_by_round[r] = f1

    round_table = pd.DataFrame(round_rows)

    # ---------- summary (Result 5.2 table) ----------
    sc_f1 = f1_by_round.get(0, 0.0)
    best_round = max(f1_by_round, key=f1_by_round.get) if f1_by_round else 0
    best_f1 = f1_by_round[best_round] if f1_by_round else 0.0

    summary_table = pd.DataFrame([{
        "SC_F1": sc_f1,
        "Best_F1": best_f1,
        "Best_Round": best_round,
        "Delta_F1": best_f1 - sc_f1
    }])

    return round_table, summary_table


# ---------- usage ----------
# 假设你的原始数据已经 load 成 python list，叫 data
# data = json.load(...)

round_table, summary_table = debate_results_from_raw(data)

print("===== Debate Performance by Round =====")
print(round_table)

print("\n===== Debate Summary (Result 5.2) =====")
print(summary_table)
```
